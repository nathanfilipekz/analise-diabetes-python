```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (8, 5)

df = pd.read_csv('diabetes.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
colunas = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
df[colunas] = df[colunas].replace(0, np.nan)
df[colunas] = df[colunas].fillna(df[colunas].median())
df.describe()
```

```python
ax = sns.countplot(x='Outcome', data=df, hue='Outcome', palette=['#4C72B0', '#DD8452'], legend=False)
plt.title('Distribuição de Diabetes', fontsize=14, fontweight='bold')
plt.xlabel('Diagnóstico')
plt.ylabel('Número de Pacientes')
ax.set_xticks([0, 1])
ax.set_xticklabels(['Não tem', 'Tem diabetes'])
ax.bar_label(ax.containers[0])
plt.show()
```

```python
ax = sns.boxplot(x='Outcome', y='Glucose', data=df, hue='Outcome', palette=['#4C72B0', '#DD8452'], legend=False)
plt.title('Níveis de Glicose por Diagnóstico', fontsize=14, fontweight='bold')
plt.xlabel('Diagnóstico')
plt.ylabel('Nível de Glicose')
ax.set_xticks([0, 1])
ax.set_xticklabels(['Não tem', 'Tem diabetes'])
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5, square=True)
plt.title('Mapa de Correlação entre as Variáveis', fontsize=14, fontweight='bold')
plt.show()
```
